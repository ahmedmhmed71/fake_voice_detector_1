```python
import librosa
import numpy as np
```

```python
SR = 16000        # Sample rate
N_MELS = 128      # عدد الـ Mel bands
MAX_FRAMES = 256  # عدد الفريمات
N_FFT = 1024      # حجم FFT
HOP_LENGTH = 256  # خطوة الإزاحة بين الفريمات
```

```python
def extract_melspec(file_path,
                    sr=SR,
                    n_mels=N_MELS,
                    n_fft=N_FFT,
                    hop_length=HOP_LENGTH,
                    max_frames=MAX_FRAMES):
    # 1) تحميل الصوت
    y, sr = librosa.load(file_path, sr=sr)

    # 2) حساب Mel-Spectrogram (طاقة)
    S = librosa.feature.melspectrogram(
        y=y,
        sr=sr,
        n_fft=n_fft,
        hop_length=hop_length,
        n_mels=n_mels
    )  # شكلها (n_mels, frames)

    # 3) تحويلها إلى dB (log-mel) عشان تكون مناسبة لـ CNN
    S_db = librosa.power_to_db(S, ref=np.max)  # قيم تقريبًا من -80 إلى 0 dB

    # 4) تثبيت عدد الفريمات
    if S_db.shape[1] < max_frames:
        pad_width = max_frames - S_db.shape[1]
        S_db = np.pad(S_db, ((0, 0), (0, pad_width)), mode='constant', constant_values=-80.0)
    else:
        S_db = S_db[:, :max_frames]

    return S_db  # (n_mels, max_frames)
```

```python
import os

DATASET_DIR = "dataset"
REAL_DIR = "/home/ahmed/Desktop/Codex/fakevoice_project/Dataset/all" # أصوات حقيقية
FAKE_DIR = "/home/ahmed/Desktop/Codex/fakevoice_project/outputs_eleven_wav"   # أصوات مولدة

X = []
y = []

# الأصوات الحقيقية (label = 0)
for fname in os.listdir(REAL_DIR):
    if fname.lower().endswith(".wav"):
        path = os.path.join(REAL_DIR, fname)
        spec = extract_melspec(path)
        X.append(spec)
        y.append(0)

# الأصوات المولدة (label = 1)
for fname in os.listdir(FAKE_DIR):
    if fname.lower().endswith(".wav"):
        path = os.path.join(FAKE_DIR, fname)
        spec = extract_melspec(path)
        X.append(spec)
        y.append(1)

X = np.array(X)  # (num_samples, N_MELS, MAX_FRAMES)
y = np.array(y)

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.preprocessing import StandardScaler

num_samples, n_mels, max_frames = X.shape

# flatten لكل عينة
X_reshaped = X.reshape(num_samples, -1)  # (num_samples, n_mels * max_frames)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_reshaped)

# نرجعه لشكل صورة
X_scaled = X_scaled.reshape(num_samples, n_mels, max_frames)

# إضافة قناة واحدة للـ CNN
X_scaled = X_scaled[..., np.newaxis]  # (num_samples, n_mels, max_frames, 1)

print("X_scaled shape:", X_scaled.shape)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("Train:", X_train.shape, y_train.shape)
print("Test :", X_test.shape, y_test.shape)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models

input_shape = (N_MELS, MAX_FRAMES, 1)

model = models.Sequential([
    layers.Conv2D(1, (2, 2), activation='relu', padding='same', input_shape=input_shape),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(2, (2, 2), activation='relu', padding='same'),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(4, (2, 2), activation='relu', padding='same'),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.5),

    layers.Dense(1, activation='sigmoid')  # تصنيف ثنائي: حقيقي/مولّد
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
history = model.fit(
    X_train, y_train,
    validation_split=0.2,
    epochs=20,
    batch_size=32
)
```

```python
from sklearn.metrics import confusion_matrix, classification_report

# توقعات على بيانات الاختبار
y_pred_prob = model.predict(X_test)
y_pred = (y_pred_prob >= 0.5).astype(int)

print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred, target_names=["REAL", "FAKE"]))
```

```python
test_loss, test_acc = model.evaluate(X_test, y_test)
print("Test Loss:", test_loss)
print("Test Accuracy:", test_acc)
```

```python
def predict_file(file_path):
    spec = extract_melspec(file_path)  # (N_MELS, MAX_FRAMES)
    
    # نفس التطبيع المستخدم في التدريب
    spec_flat = spec.reshape(1, -1)
    spec_scaled = scaler.transform(spec_flat)
    spec_scaled = spec_scaled.reshape(1, N_MELS, MAX_FRAMES, 1)

    prob = model.predict(spec_scaled)[0][0]
    label = "FAKE (مولّد)" if prob >= 0.5 else "REAL (حقيقي)"
    print(f"Probability: {prob:.3f} → {label}")

# مثال:
# predict_file("test_example.wav")
```
